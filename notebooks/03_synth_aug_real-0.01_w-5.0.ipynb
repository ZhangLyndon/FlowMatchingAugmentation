We evaluate classification performance under synthetic data augmentation. 1% of the training set is combined with synthetic samples equivalent to 100% of the original dataset (6,000 per class), yielding a training set equal to 101% of the original size. Synthetic samples are generated with a guidance scale of $w = 5.0$, and performance is evaluated using both accuracy and the macro $\mathsf F_1$ score.

```python
# Configure augmentation evaluation pipeline
augmentation_args = argparse.Namespace(data_root = "./data",
                                       batch_size = 16,
                                       num_workers = 0,
                                       epochs = 25,
                                       lr = 0.001,
                                       weight_decay = 1e-4,
                                       step_size = 15,
                                       gamma = 0.1,
                                       synthetic_data_dir = "./images",
                                       real_ratio = 0.01,
                                       classification_dir = "./results/classification",
                                       augmentation_dir = "./results/augmentation",
                                       checkpoint_dir = "./checkpoints",
                                       save_interval = 20,
                                       seed = 42)

# Create directory to store synthetic augmentation results
os.makedirs(augmentation_args.augmentation_dir, exist_ok = True)

# Set random seed for reproducibility
torch.manual_seed(augmentation_args.seed)
np.random.seed(augmentation_args.seed)
```

```python
guidance_scale = 5.0
evaluator = SyntheticAugmentationEvaluator(augmentation_args, guidance_scale)
evaluator.run_low_data_experiments(augmentation_args.real_ratio, True)
```

Upon applying augmentation to the extremely low-data training split (i.e., 1% of the training set), the model achieves a classification accuracy of $81.29\%$, a macro $\mathsf F_1$ score of $0.8138$, and an optimal validation (cross-entropy) loss of $0.6810$ at epoch 19.
